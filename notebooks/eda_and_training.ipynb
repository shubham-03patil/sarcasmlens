```python
import pandas as pd

df = pd.read_json("../data/Sarcasm_Headlines_Dataset_v2.json", lines=True)
df.head()
```

```python
print(df.shape)
print(df['is_sarcastic'].value_counts())
df.info()
```

```python
# Look at a few sarcastic headlines
print("SARCASTIC examples:")
print(df[df['is_sarcastic'] == 1]['headline'].sample(5, random_state=42).to_list())

print("\nNOT SARCASTIC examples:")
print(df[df['is_sarcastic'] == 0]['headline'].sample(5, random_state=42).to_list())
```

```python
df['headline_length'] = df['headline'].apply(lambda x: len(x.split()))
df.groupby('is_sarcastic')['headline_length'].describe()
```

```python
df[df['headline_length'] > 50][['headline', 'headline_length', 'is_sarcastic']]
```
